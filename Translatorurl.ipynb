```python
from types import resolve_bases
import requests
from bs4 import BeautifulSoup

def extract_text_from_url(url):
  response = requests.get(url)

  if response.status_code == 200:
    soup = BeautifulSoup(response.text, 'html.parser')
    for script_or_style in soup(['script','style']):
      script_or_style.decompose()

    texto = soup.get_text(separator= ' ')
    #Limpar texto
    linhas = (line.strip() for line in texto.splitlines())
    parts = (phrase.strip() for line in linhas for phrase in line.split(" "))
    texto_limpo = '\n'.join(part for part in parts if part)
    return texto_limpo


  else:
    print(f"Failed to fetch the URL. Status code: {resolve_status_code}")
    return None

  soup = BeautifulSoup(response.text, 'html.parser')
  text = soup.get_text()
  return text

extract_text_from_url('https://dev.to/borisgigovic/azure-openai-in-production-technical-insights-from-the-trenches-50o2')
```

```python
from langchain_openai.chat_models.azure import AzureChatOpenAI
from langchain.schema import SystemMessage, HumanMessage

client = AzureChatOpenAI(
    azure_endpoint="seu_ponto_de_extremidade_do_Azure",
    api_key="Sua_chave1_do_Azure",
    api_version="2024-02-15-preview",
    deployment_name="gpt-4o-mini",
    max_retries=0
)

def translate_article(text, lang):
    messages = [
        SystemMessage(content="Você atua como tradutor de textos."),
        HumanMessage(content=f"Traduza o seguinte texto para o idioma {lang} e responda em markdown:\n\n{text}")
    ]
    response = client.invoke(messages)
    print(response.content)
    return response.content

# Chamada correta da função (fora da definição)
translate_article("Every time a new technology shows up, someone loudly declares it's the end of the world.", "português")
```

```python
url = 'https://dev.to/besworks/will-ai-ruin-society-2e02'
text = extract_text_from_url(url)
article = translate_article(text, 'pt-br')

print(article)
```
